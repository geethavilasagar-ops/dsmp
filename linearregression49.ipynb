```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
df.shape
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package (in Lpa)')

```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

Breaking it down
df.iloc[rows, columns]

First part before comma → which rows

Second part after comma → which columns
​

For x = df.iloc[:,0:1]
: → all rows.

0:1 → columns from position 0 up to (but not including) 1, so only the first column.

Result: x is a DataFrame containing all rows, only the first column (e.g., cgpa in your file).
​

For y = df.iloc[:,-1]
: → all rows.

-1 → the last column by position.

Result: y is a Series containing all rows of the last column

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
X_test
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('package(in lpa)')
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred = lr.predict(X_test)
```

```python
y_test.values
```

```python
print("MAE",mean_absolute_error(y_test,y_pred))
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```

```python
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

 If a number is between 0 and 1 (like 0.121), its square root will actually be larger than the number itself.

```python
print("MSE",r2_score(y_test,y_pred))
r2 = r2_score(y_test,y_pred)
```

It measures how well the model's predictions (y_pred) match the actual values (y_test). The r2_score function from scikit-learn is used for this purpose.

```python
X_test.shape
```
