```python
import random
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

def pltPer(X, y, W, it = -1): # Fixed spacing for it = -1
    title = "Perceptron Learning Algorithm" # Added indentation
    if it > 0: # Fixed @ to 0
        title = title + " After Update #" + str(it)

    f = plt.figure()
    for n in range(len(y)):
        if y[n] == 1:
            plt.plot(X[n,1],X[n,2],'r+')
        else:
            plt.plot(X[n,1],X[n,2],'k_')

    # Plot the decision boundary
    if np.isclose(W[2], 0):
        if not np.isclose(W[1], 0):
            # W[2] is zero, W[1] is not zero: vertical line x1 = -W[0]/W[1]
            plt.axvline(x=-W[0]/W[1], color='k', linestyle='-')
        # If both W[2] and W[1] are zero, and W[0] is also zero (initial state),
        # there's no line to plot, so we do nothing.
    else:
        # W[2] is not zero: standard line x2 = m*x1 + b
        m, b = -W[1]/W[2], -W[0]/W[2]
        l = np.linspace(min(X[:,1] ), max(X[ :,1]))
        plt.plot(l, m*l+b, 'k-')

    plt.axis( 'tight')
    plt.xlim(min(X[:,1])-0.1, max(X[:,1])+0.1)
    plt.ylim(min(X[:,2])-0.1, max(X[ :,2])+0.1) # Fixed 8.1 to 0.1

    plt.ylabel("$x_2$")
    plt.title(title)
    plt.savefig('ch.5.PLA.'+str(it)+'.png', dpi=350, bbox_inches='tight')
    plt.savefig('ch.5.PLA.'+str(it)+'.eps', dpi=350, bbox_inches='tight')
    plt.show()
```

```python
def classification_error(w, X, y):
    err_cnt = 0
    N = len(X) # Fixed Len to len
    for n in range(N):
        s_prediction = np.sign(w.T.dot(X[n])) # s_prediction is the predicted sign
        if s_prediction != y[n]: # Compare prediction with actual label
            err_cnt += 1
    return err_cnt

def choose_miscl_point(w, X, y):
    # Iterates through points in a random order and returns the first misclassified point and its label.
    # Assumes that classification_error(w, X, y) != 0, so a misclassified point will be found.
    for idx in np.random.permutation(len(X)):
        if np.sign(w.T.dot(X[idx])) != y[idx]:
            return X[idx], y[idx]
    return None, None # Should not be reached if there are misclassified points

# main
N = 100
random.seed(a=7) # Fixed space
np.random.seed(7)

X, y = make_classification(n_samples=N, n_features=2, n_classes=2,
                           n_informative=2, n_redundant=0, n_repeated=0, # Fixed n_redundant from 8 to 0, added comma
                           n_clusters_per_class=1, class_sep=1.2,
                           random_state=5)

y[y == 0] = -1

X_train = np.append(np.ones((N,1)), X, 1) # add a column of ones

# initialize the weights to zeros
w = np.zeros(X_train.shape[1])
it = 0
pltPer(X_train,y,w) # initial solution (likely bad)

# Iterate until all points are correctly classified
while classification_error(w, X_train, y) != 0:
    it += 1
    # Pick random misclassified point
    misclassified_point, misclassified_label = choose_miscl_point(w, X_train, y)
    # Update weights
    w += misclassified_label * misclassified_point # Corrected perceptron update
    if it > 1000:
        break
    pltPer(X_train,y,w, it)
```
